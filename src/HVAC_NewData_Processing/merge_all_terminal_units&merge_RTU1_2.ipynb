```python
import pandas as pd
import os
```

```python
# Configuration
data_folder = '../../HVAC_NewData'  
output_file = '../../outputs/all_terminal_units.csv'  # Output filename
rtu_merged_file = '../../outputs/RTU_merged.csv' 
connection_table = '../../outputs/connection_mapping_analysis/final_connection_table.csv' 
```

```python
# Step 1: Load the connection table
connection_df = pd.read_csv(connection_table)
# Create mapping dictionary: {'A': 'RTU_1', 'AA': 'RTU_1', ...}
vav_rtu_mapping = dict(zip(connection_df['VAV'], connection_df['RTU']))
```

```python
# Step 2: Initialize an empty list to store all DataFrames
all_terminal_units = []
```

```python
# Step 3: Process each terminal unit file
for filename in os.listdir(data_folder):
    if filename.endswith('.csv') and filename not in ['RTU_1.csv', 'RTU_2.csv', 'RTU_3.csv']:
        # Extract VAV_ID from filename (remove .csv)
        vav_id = filename.replace('.csv', '')
        
        try:
            # Load the terminal unit data
            filepath = os.path.join(data_folder, filename)
            df = pd.read_csv(filepath)
            
            # Add identifier columns
            df['VAV_ID'] = vav_id
            df['RTU'] = vav_rtu_mapping.get(vav_id, 'UNKNOWN')  # Get RTU from mapping
            
            # Reorder columns to have IDs first
            cols = ['VAV_ID', 'RTU'] + [col for col in df.columns if col not in ['VAV_ID', 'RTU']]
            df = df[cols]
            
            # Add to our collection
            all_terminal_units.append(df)
            
        except Exception as e:
            print(f"Error processing {filename}: {str(e)}")
            continue
```

```python
# Step 4: Combine all DataFrames
if all_terminal_units:
    integrated_data = pd.concat(all_terminal_units, ignore_index=True)
    
    # Save the integrated data
    integrated_data.to_csv(output_file, index=False)
    print(f"Successfully integrated {len(all_terminal_units)} terminal units. Saved to {output_file}")
    
    # Display sample of the result
    print("\nSample of integrated data:")
    display(integrated_data.head())
else:
    print("No terminal unit files were processed.")
```

```python
# Step 1: Load both RTU files
rtu1 = pd.read_csv(os.path.join(data_folder, 'RTU_1.csv'))
rtu2 = pd.read_csv(os.path.join(data_folder, 'RTU_2.csv'))
```

```python
# Step 2: Add identifier columns
rtu1['RTU_ID'] = 'RTU_1'
rtu2['RTU_ID'] = 'RTU_2'
```

```python
# Step 3: Combine the DataFrames
rtu_merged = pd.concat([rtu1, rtu2], ignore_index=True)
```

```python
# Step 4: Reorder columns to have RTU_ID first
cols = ['RTU_ID'] + [col for col in rtu_merged.columns if col != 'RTU_ID']
rtu_merged = rtu_merged[cols]
```

```python
# Step 5: Save the merged data
rtu_merged.to_csv(rtu_merged_file, index=False)
print(f"Successfully merged RTU files. Saved to {rtu_merged_file}")
# Display sample of the result
print("\nSample of merged RTU data:")
display(rtu_merged.head())
```
