```python
import pandas as pd
import numpy as np
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
# Load the integrated data
df = pd.read_csv('../../outputs/all_terminal_units.csv')
```

```python
# ======================
# 1. Time Format Standardization
# ======================
print("Original time format samples:")
print(df['time'].head(3))

# Convert to datetime with error handling
df['time'] = pd.to_datetime(df['time'], errors='coerce')

# Check for failed conversions
time_nulls = df['time'].isnull().sum()
print(f"\n{time_nulls} rows with invalid time formats will be dropped")
df = df.dropna(subset=['time'])

# Sort by time for each VAV unit
df = df.sort_values(['VAV_ID', 'time'])
```

```python
# ======================
# 2. Column Consistency Check
# ======================
# List of expected columns (from your description)
expected_columns = [
    'VAV_ID', 'RTU', 'time', 'AirFlow', 'Air_Flow_Diff', 'ClgPct', 'DmprPos', 
    'DschAirTemp', 'EffAirflowSp', 'EffOcc', 'EffSp', 'HtgPct', 'OccClgSp', 
    'OccCmd', 'OccHtgSp', 'RmSp', 'RmTemp', 'Room_Temperature_Diff', 'SaTemp',
    'StbyClgSp', 'StbyHtgSp', 'UnoccClgSp', 'UnoccHtgSp', 'VAV_Temperature_Diff',
    'oppMode', 'RmCo2', 'SuplFanCmd', 'SuplFanState', 'AirflowSpRht', 'SaTmp',
    'Cooling_Stg_Cmd', 'Heating_Stg_Cmd'
]

# Add missing columns with NaN values
for col in expected_columns:
    if col not in df.columns:
        df[col] = np.nan

# Reorder columns
df = df[expected_columns]
```

```python
# ======================
# 3. Missing Value Analysis
# ======================
print("\nMissing value analysis:")
missing_stats = df.isnull().mean().sort_values(ascending=False)
print(missing_stats[missing_stats > 0])

# Handle missing values based on column type
for col in df.columns:
    if col in ['VAV_ID', 'RTU', 'time']:
        continue  # These should have no missing values after previous steps
        
    elif df[col].dtype in ['float64', 'int64']:
        # For numerical columns, fill with median of the same VAV unit
        df[col] = df.groupby('VAV_ID')[col].transform(
            lambda x: x.fillna(x.median()))
            
    elif df[col].dtype == 'object':
        # For categorical columns, fill with mode
        df[col] = df.groupby('VAV_ID')[col].transform(
            lambda x: x.fillna(x.mode()[0] if len(x.mode()) > 0 else 'UNKNOWN'))
```

```python
# ======================
# 4. Sensor Anomaly Detection
# ======================
# Define reasonable ranges for key metrics (adjust based on domain knowledge)
valid_ranges = {
    'AirFlow': (0, 1500),          # CFM
    'RmTemp': (50, 90),            #°F
    'DschAirTemp': (45, 120),      #°F
    'ClgPct': (0, 100),            #%
    'HtgPct': (0, 100),            #%
    'DmprPos': (0, 100),           #%
    'RmCo2': (300, 2000),          #ppm
}

# Flag outliers
for col, (min_val, max_val) in valid_ranges.items():
    if col in df.columns:
        df[f'{col}_outlier'] = ~df[col].between(min_val, max_val)

# Count outliers per VAV unit
outlier_cols = [col for col in df.columns if '_outlier' in col]
df['total_outliers'] = df[outlier_cols].sum(axis=1)

print("\nTop VAV units with most outliers:")
print(df.groupby('VAV_ID')['total_outliers'].sum().sort_values(ascending=False).head(10))
```

```python
# ======================
# 5. Time Series Consistency
# ======================
# Check time intervals for each VAV
time_analysis = df.groupby('VAV_ID')['time'].agg([
    ('start', 'min'),
    ('end', 'max'),
    ('count', 'count'),
    ('interval', lambda x: x.diff().median())
]).reset_index()

print("\nTime coverage analysis:")
print(time_analysis)
```

```python
# ======================
# 6. Save Cleaned Data
# ======================
df.to_csv('../../outputs/cleaned_terminal_units.csv', index=False)
print("\nData cleaning complete. Saved to cleaned_terminal_units.csv")

# Display cleaned data sample
print("\nCleaned data sample:")
display(df.head())
```

```python
# Load RTU merged data
rtu_df = pd.read_csv('../../outputs/RTU_merged.csv')
```

```python
# =====================================
# 1. Initial Data Inspection
# =====================================
print("Initial Data Overview:")
print(f"Shape: {rtu_df.shape}")
print("\nData Types:")
print(rtu_df.dtypes)
print("\nMissing Values per Column:")
print(rtu_df.isnull().mean().sort_values(ascending=False))
```

```python
# =====================================
# 2. Time Column Standardization
# =====================================
# Convert to datetime with error handling
rtu_df['time'] = pd.to_datetime(rtu_df['time'], errors='coerce')

# Check for invalid timestamps
invalid_time = rtu_df['time'].isnull().sum()
print(f"\nDropping {invalid_time} rows with invalid timestamps")
rtu_df = rtu_df.dropna(subset=['time'])

# Sort by RTU_ID and time
rtu_df = rtu_df.sort_values(['RTU_ID', 'time'])
```

```python
# =====================================
# 3. Column-Specific Cleaning
# =====================================
# Define valid ranges for key parameters (adjust based on domain knowledge)
valid_ranges = {
    'BldgStaticPres': (0, 10),          # Typical building static pressure (in. w.g.)
    'ClgPct': (0, 100),                 # Cooling percentage (%)
    'HtgPct': (0, 100),                 # Heating percentage (%)
    'OutAirTemp': (-20, 120),           # Outdoor air temperature (°F)
    'RmTemp': (50, 90),                 # Room temperature (°F)
    'SuplFanSpd': (0, 100),             # Supply fan speed (%)
    'OutAirFlow': (0, 50000),           # Outdoor airflow (CFM)
    'SuplAirTemp': (45, 120)            # Supply air temperature (°F)
}

# Clean each column
for col, (min_val, max_val) in valid_ranges.items():
    if col in rtu_df.columns:
        # Replace values outside range with NaN
        rtu_df[col] = rtu_df[col].where(rtu_df[col].between(min_val, max_val), np.nan)
        
        # Create outlier flag column
        rtu_df[f'{col}_outlier'] = ~rtu_df[col].between(min_val, max_val)

# Handle binary/categorical columns
binary_cols = ['EconEna', 'EconEnaSp', 'FltAlm', 'HtgAlm', 'OccCmd', 'OccStatus', 
               'SuplFanAlm', 'SuplFanStatus', 'SuplFanSts', 'oppMode']

for col in binary_cols:
    if col in rtu_df.columns:
        # Standardize to 0/1 if needed
        if rtu_df[col].nunique() == 2:
            modes = rtu_df[col].mode()
            if len(modes) > 0:
                rtu_df[col] = np.where(rtu_df[col] == modes[0], 0, 1)
```

```python
# =====================================
# 4. Missing Value Treatment 
# =====================================
print("\nMissing value analysis before treatment:")
print(rtu_df.isnull().mean().sort_values(ascending=False))

# Handle missing values (skip RTU_ID and time)
cols_to_clean = [col for col in rtu_df.columns if col not in ['RTU_ID', 'time']]

for col in cols_to_clean:
    if rtu_df[col].dtype in ['float64', 'int64']:
        # For numerical columns, fill with median of the same RTU unit
        rtu_df[col] = rtu_df.groupby('RTU_ID')[col].transform(
            lambda x: x.fillna(x.median()))
            
    elif rtu_df[col].dtype == 'object':
        # For categorical columns, fill with mode
        rtu_df[col] = rtu_df.groupby('RTU_ID')[col].transform(
            lambda x: x.fillna(x.mode()[0] if len(x.mode()) > 0 else 'UNKNOWN'))

print("\nMissing value analysis after treatment:")
print(rtu_df.isnull().mean().sort_values(ascending=False))
```

```python
# =====================================
# 5. Time Series Consistency Check
# =====================================
# Check time intervals for each RTU
time_stats = rtu_df.groupby('RTU_ID')['time'].agg([
    ('start_time', 'min'),
    ('end_time', 'max'),
    ('record_count', 'count'),
    ('avg_interval', lambda x: x.diff().median())
]).reset_index()

print("\nTime Coverage Statistics:")
print(time_stats)
```

```python
# =====================================
# 6. Save Cleaned Data
# =====================================
rtu_df.to_csv('../../outputs/cleaned_RTU_merged.csv', index=False)
print("\nCleaning complete. Saved to cleaned_RTU_merged.csv")

# Display sample of cleaned data
print("\nCleaned Data Sample:")
display(rtu_df.head())
```

```python
# Load RTU_3 data
rtu3_df = pd.read_csv('../../HVAC_NewData/RTU_3.csv')
# =====================================
# 1. Initial Data Inspection
# =====================================
print("Initial Data Overview:")
print(f"Shape: {rtu3_df.shape}")
print("\nData Types:")
print(rtu3_df.dtypes)
print("\nMissing Values per Column:")
print(rtu3_df.isnull().mean().sort_values(ascending=False))
# =====================================
# 2. Time Column Standardization
# =====================================
# Convert to datetime with error handling
rtu3_df['time'] = pd.to_datetime(rtu3_df['time'], errors='coerce')

# Check for invalid timestamps
invalid_time = rtu3_df['time'].isnull().sum()
print(f"\nDropping {invalid_time} rows with invalid timestamps")
rtu3_df = rtu3_df.dropna(subset=['time'])

# Sort by RTU_ID and time
rtu3_df = rtu3_df.sort_values(['time'])
# =====================================
# 3. Column-Specific Cleaning
# =====================================
# Define valid ranges for key parameters (adjust based on domain knowledge)
valid_ranges = {
    'BldgStaticPres': (0, 10),  # Typical building static pressure (in. w.g.)
    'ClgPct': (0, 100),  # Cooling percentage (%)
    'HtgPct': (0, 100),  # Heating percentage (%)
    'OutAirTemp': (-20, 120),  # Outdoor air temperature (°F)
    'RmTemp': (50, 90),  # Room temperature (°F)
    'SuplFanSpd': (0, 100),  # Supply fan speed (%)
    'OutAirFlow': (0, 50000),  # Outdoor airflow (CFM)
    'SuplAirTemp': (45, 120)  # Supply air temperature (°F)
}

# Clean each column
for col, (min_val, max_val) in valid_ranges.items():
    if col in rtu3_df.columns:
        # Replace values outside range with NaN
        rtu3_df[col] = rtu3_df[col].where(rtu3_df[col].between(min_val, max_val), np.nan)

        # Create outlier flag column
        rtu3_df[f'{col}_outlier'] = ~rtu3_df[col].between(min_val, max_val)

# Handle binary/categorical columns
binary_cols = ['EconEna', 'EconEnaSp', 'FltAlm', 'HtgAlm', 'OccCmd', 'OccStatus',
               'SuplFanAlm', 'SuplFanStatus', 'SuplFanSts', 'oppMode']

for col in binary_cols:
    if col in rtu3_df.columns:
        # Standardize to 0/1 if needed
        if rtu3_df[col].nunique() == 2:
            modes = rtu3_df[col].mode()
            if len(modes) > 0:
                rtu3_df[col] = np.where(rtu3_df[col] == modes[0], 0, 1)
# =====================================
# 4. Missing Value Treatment
# =====================================
print("\nMissing value analysis before treatment:")
print(rtu3_df.isnull().mean().sort_values(ascending=False))

# Handle missing values (skip time column)
cols_to_clean = [col for col in rtu3_df.columns if col != 'time']

for col in cols_to_clean:
    if rtu3_df[col].dtype in ['float64', 'int64']:
        # For numerical columns, fill with median
        rtu3_df[col] = rtu3_df[col].fillna(rtu3_df[col].median())
            
    elif rtu3_df[col].dtype == 'object':
        # For categorical columns, fill with mode
        mode_val = rtu3_df[col].mode()[0] if len(rtu3_df[col].mode()) > 0 else 'UNKNOWN'
        rtu3_df[col] = rtu3_df[col].fillna(mode_val)

print("\nMissing value analysis after treatment:")
print(rtu3_df.isnull().mean().sort_values(ascending=False))

# =====================================
# 5. Time Series Consistency Check
# =====================================
# Calculate time statistics
time_stats = pd.DataFrame({
    'start_time': [rtu3_df['time'].min()],
    'end_time': [rtu3_df['time'].max()],
    'record_count': [len(rtu3_df)],
    'avg_interval': [rtu3_df['time'].diff().median()]
})

print("\nTime Coverage Statistics:")
print(time_stats)

# =====================================
# 6. Save Cleaned Data
# =====================================
rtu3_df.to_csv('../../outputs/cleaned_RTU_3.csv', index=False)
print("\nCleaning complete. Saved to cleaned_RTU_merged.csv")

# Display sample of cleaned data
print("\nCleaned Data Sample:")
display(rtu3_df.head())
```
