```python
# =============================================
# 2. Energy Efficiency Metrics Calculation
# =============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# 1. Load previous analysis results
input_file = "../../outputs/HVAC_NewData_Analysis_Results/basic_statistical_analysis_20250404_202959.xlsx"
vav_stats = pd.read_excel(input_file, sheet_name='VAV_Statistics', index_col='VAV_ID')
```

```python
# 2. Load original cleaned data for raw metrics
vav_df = pd.read_csv('../../outputs/HVAC_NewData_Analysis_Data/cleaned_terminal_units.csv', parse_dates=['time'])

# =============================================
# 2.1 Energy Usage Intensity (EUI) Proxy Calculation
# =============================================
def calculate_eui(df, area_col=None):
    """Calculate energy usage intensity proxy"""
    # If building area data available (replace with actual column name)
    if area_col and area_col in df.columns:
        df['EUI'] = df['AirFlow'] * df['ClgPct'] / df[area_col]  # kBTU/hr/sqft
    else:
        # Normalized proxy (dimensionless)
        df['EUI_proxy'] = (df['AirFlow'] * df['ClgPct'] / df['EffAirflowSp']).fillna(0)
    return df

vav_df = calculate_eui(vav_df)
```

```python
# =============================================
# 2.2 System Efficiency Metrics
# =============================================
def calculate_efficiency_metrics(df):
    """Calculate key efficiency indicators"""
    # Airflow efficiency
    df['Airflow_Ratio'] = df['AirFlow'] / df['EffAirflowSp']
    
    # Temperature control accuracy
    df['Temp_Control_Error'] = (df['RmTemp'] - df['RmSp']).abs()
    
    # Mode efficiency (cooling vs heating)
    df['Mode_Conflict'] = np.where(
        (df['ClgPct'] > 20) & (df['HtgPct'] > 20), 1, 0)
    
    # Damper efficiency
    df['Damper_Efficiency'] = np.where(
        df['DmprPos'] > 0,
        df['AirFlow'] / (df['DmprPos'] * df['EffAirflowSp']),
        np.nan
    )
    return df

vav_df = calculate_efficiency_metrics(vav_df)
```

```python
# =============================================
# 2.3 Benchmarking Analysis (Corrected)
# =============================================
def benchmark_units(df, stats_df):
    """Compare units against statistical baselines"""
    # Calculate required statistics if not present
    required_stats = ['AirFlow_mean', 'AirFlow_std']
    
    # Check if EUI_proxy exists in original data
    if 'EUI_proxy' in df.columns:
        # Calculate EUI statistics if not in stats_df
        if 'EUI_proxy_mean' not in stats_df.columns:
            eui_stats = df.groupby('VAV_ID')['EUI_proxy'].agg(['mean', 'std'])
            eui_stats.columns = ['EUI_proxy_mean', 'EUI_proxy_std']
            stats_df = stats_df.join(eui_stats)
        required_stats.extend(['EUI_proxy_mean', 'EUI_proxy_std'])
    
    # Merge with stats data
    merged = df.join(stats_df[required_stats], on='VAV_ID')
    
    # Flag outliers
    for metric in ['AirFlow'] + (['EUI_proxy'] if 'EUI_proxy' in df.columns else []):
        mean_col = f'{metric}_mean'
        std_col = f'{metric}_std'
        
        if mean_col in merged.columns and std_col in merged.columns:
            merged[f'{metric}_benchmark'] = np.where(
                (merged[metric] > merged[mean_col] + 2*merged[std_col]) |
                (merged[metric] < merged[mean_col] - 2*merged[std_col]),
                'Outlier', 'Normal'
            )
    
    return merged

# Apply benchmarking
vav_df = benchmark_units(vav_df, vav_stats)

# Verify results
print("Benchmarking columns added:", 
      [col for col in vav_df.columns if '_benchmark' in col])
```

```python
# =============================================
# 2.4 Corrected Visualization
# =============================================
plt.figure(figsize=(15, 10))

# 1. First make sure all the required columns are present
required_columns = {
    'EUI_proxy': 'Normalized EUI',
    'Airflow_Ratio': 'AirFlow/EffAirflowSp Ratio',
    'Temp_Control_Error': 'Temperature Control Error',
    'Mode_Conflict': 'Mode Conflict'
}

# Dynamically create missing columns
if 'Airflow_Ratio' not in vav_df.columns and all(col in vav_df.columns for col in ['AirFlow', 'EffAirflowSp']):
    vav_df['Airflow_Ratio'] = vav_df['AirFlow'] / vav_df['EffAirflowSp']

if 'Temp_Control_Error' not in vav_df.columns and all(col in vav_df.columns for col in ['RmTemp', 'RmSp']):
    vav_df['Temp_Control_Error'] = (vav_df['RmTemp'] - vav_df['RmSp']).abs()

if 'Mode_Conflict' not in vav_df.columns and all(col in vav_df.columns for col in ['ClgPct', 'HtgPct']):
    vav_df['Mode_Conflict'] = np.where(
        (vav_df['ClgPct'] > 20) & (vav_df['HtgPct'] > 20), 1, 0)

# 2. 绘制图表
try:
    # Plot 1: EUI Distribution
    plt.subplot(2, 2, 1)
    if 'EUI_proxy' in vav_df.columns:
        sns.boxplot(data=vav_df, x='RTU', y='EUI_proxy')
        plt.title('Normalized EUI Distribution by RTU')
        plt.ylim(0, 2)
    else:
        plt.text(0.5, 0.5, 'EUI_proxy data not available', ha='center')
        plt.title('Data Missing')

    # Plot 2: Airflow Efficiency
    plt.subplot(2, 2, 2)
    if all(col in vav_df.columns for col in ['DmprPos', 'Airflow_Ratio']):
        sns.scatterplot(data=vav_df, x='DmprPos', y='Airflow_Ratio', hue='RTU')
        plt.axhline(1, color='r', linestyle='--')
        plt.title('Damper Position vs Airflow Efficiency')
    else:
        missing = [col for col in ['DmprPos', 'Airflow_Ratio'] if col not in vav_df.columns]
        plt.text(0.5, 0.5, f'Missing columns: {", ".join(missing)}', ha='center')
        plt.title('Data Missing')

    # Plot 3: Temperature Control
    plt.subplot(2, 2, 3)
    if 'Temp_Control_Error' in vav_df.columns:
        vav_df['Temp_Control_Error'].plot(kind='hist', bins=30)
        plt.title('Temperature Control Error Distribution')
    else:
        plt.text(0.5, 0.5, 'Temp_Control_Error data not available', ha='center')
        plt.title('Data Missing')

    # Plot 4: Mode Conflicts
    plt.subplot(2, 2, 4)
    if 'Mode_Conflict' in vav_df.columns:
        vav_df['Mode_Conflict'].value_counts().plot(kind='pie', autopct='%1.1f%%')
        plt.title('Simultaneous Heating/Cooling Occurrence')
    else:
        plt.text(0.5, 0.5, 'Mode_Conflict data not available', ha='center')
        plt.title('Data Missing')

    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"Plotting error: {str(e)}")
```

```python
# =============================================
# 2.5 Save Results
# =============================================
output_dir = Path("../../outputs/HVAC_NewData_Analysis_Results")
output_dir.mkdir(exist_ok=True)

# Save efficiency metrics
efficiency_file = output_dir / "2.energy_efficiency_metrics.csv"
vav_df.to_csv(efficiency_file, index=False)

# Generate summary report
summary = vav_df.groupby('RTU').agg({
    'EUI_proxy': ['mean', 'std'],
    'Airflow_Ratio': lambda x: (x < 0.8).mean(),
    'Temp_Control_Error': ['mean', 'max'],
    'Mode_Conflict': 'mean'
}).round(3)

report_file = output_dir / "2.efficiency_summary.xlsx"
summary.to_excel(report_file)

print(f"""
Analysis Completed:
- Raw efficiency metrics saved to: {efficiency_file}
- Summary report saved to: {report_file}
""")

# Display top inefficient units
print("\nTop 5 Inefficient VAVs by EUI:")
display(
    vav_df.sort_values('EUI_proxy', ascending=False)
    [['VAV_ID', 'RTU', 'EUI_proxy', 'Airflow_Ratio']]
    .head(5)
)
```
