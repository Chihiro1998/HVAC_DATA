```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from statsmodels.tsa.arima.model import ARIMA
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('../../outputs/HVAC_NewData_Analysis_Data/cleaned_terminal_units.csv', parse_dates=['time'])
rtu_df = pd.read_csv('../../outputs/HVAC_NewData_Analysis_Data/cleaned_RTU_merged.csv', parse_dates=['time'])
```

```python
# =============================================
# 第一部分：时间序列预测建模
# =============================================

def prepare_time_series_data(df, target_col='AirFlow', freq='H'):
    """准备时间序列数据"""
    ts_data = df.set_index('time').groupby('VAV_ID')[target_col].resample(freq).mean().unstack(0)
    return ts_data.fillna(method='ffill').dropna()

def train_predictive_models(ts_data):
    """训练预测模型并评估"""
    results = {}
    
    # 示例：预测第一个VAV单元
    example_vav = ts_data.columns[0]
    y = ts_data[example_vav]
    
    # 方法1：ARIMA时间序列模型
    model_arima = ARIMA(y, order=(24,1,2))  # 24小时季节性
    results['ARIMA'] = model_arima.fit()
    
    # 方法2：随机森林回归（特征工程）
    def create_features(df):
        df['hour'] = df.index.hour
        df['dayofweek'] = df.index.dayofweek
        df['lag_24'] = df[example_vav].shift(24)
        return df.dropna()
    
    X = create_features(pd.DataFrame(y))
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
    
    model_rf = RandomForestRegressor(n_estimators=100)
    model_rf.fit(X_train, y_train)
    results['RandomForest'] = model_rf
    
    # 评估指标
    print(f"\n预测模型评估 ({example_vav}):")
    for name, model in results.items():
        if name == 'ARIMA':
            pred = model.predict(start=len(y)-24, end=len(y)+24)
            plt.figure(figsize=(12,4))
            y.plot(label='Actual')
            pred.plot(label='Predicted')
            plt.title(f'ARIMA预测结果 - {example_vav}')
            plt.legend()
            plt.show()
        else:
            pred = model.predict(X_test)
            plt.figure(figsize=(12,4))
            plt.plot(y_test.index, y_test, label='Actual')
            plt.plot(y_test.index, pred, label='Predicted', alpha=0.7)
            plt.title(f'随机森林预测结果 - {example_vav}')
            plt.legend()
            plt.show()
    
    return results
# 执行预测建模
ts_data = prepare_time_series_data(df)
predictive_models = train_predictive_models(ts_data)
```

```python
# =============================================
# 第二部分：能效聚类分析
# =============================================

def prepare_clustering_features(df):
    """准备聚类特征矩阵"""
    # 基础能效指标
    features = df.groupby('VAV_ID').agg({
        'AirFlow': ['mean', 'std'],
        'RmTemp': lambda x: np.mean(np.abs(x - x.median())),
        'ClgPct': 'mean',
        'total_outliers': 'sum'
    })
    features.columns = ['AirFlow_mean', 'AirFlow_std', 'Temp_deviation', 'Cooling_ratio', 'Outlier_count']
    
    # 添加时间序列特征（日波动率）
    daily_std = df.set_index('time').groupby('VAV_ID')['AirFlow'].resample('D').std().groupby('VAV_ID').mean()
    features['Daily_volatility'] = daily_std
    
    return features.dropna()

def perform_clustering(features, n_clusters=None):
    """执行聚类分析"""
    # 标准化数据
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(features)
    
    # 确定最佳聚类数（肘部法则）
    if n_clusters is None:
        inertias = []
        for k in range(2, 10):
            kmeans = KMeans(n_clusters=k, random_state=42).fit(X_scaled)
            inertias.append(kmeans.inertia_)
        
        plt.figure(figsize=(10,5))
        plt.plot(range(2,10), inertias, marker='o')
        plt.xlabel('Number of clusters')
        plt.ylabel('Inertia')
        plt.title('Elbow Method for Optimal k')
        plt.show()
        n_clusters = int(input("根据肘部法则选择聚类数量: "))
    
    # 执行聚类
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    clusters = kmeans.fit_predict(X_scaled)
    
    # 评估轮廓系数
    print(f"轮廓系数: {silhouette_score(X_scaled, clusters):.2f}")
    
    # 分析聚类结果
    features['Cluster'] = clusters
    cluster_profiles = features.groupby('Cluster').mean()
    
    # 可视化聚类结果
    plt.figure(figsize=(12,6))
    sns.heatmap(cluster_profiles.T, cmap='YlGnBu', annot=True, fmt=".1f")
    plt.title('Cluster Profiles (Normalized)')
    plt.show()
    
    return features

# 执行聚类分析
features = prepare_clustering_features(df)
clustered_data = perform_clustering(features, n_clusters=3)  # 可指定或自动选择
```

```python
# =============================================
# 第三部分：异常单元识别
# =============================================

def identify_anomalies(clustered_data):
    """识别异常聚类组"""
    # 假设高能耗低效率的异常簇具有以下特征：
    # - 高AirFlow_mean
    # - 高Outlier_count
    # - 高Temp_deviation
    clustered_data['Anomaly_Score'] = (
        clustered_data['AirFlow_mean'] * 
        clustered_data['Outlier_count'] * 
        clustered_data['Temp_deviation']
    )
    
    # 自动检测异常阈值（IQR方法）
    q1 = clustered_data['Anomaly_Score'].quantile(0.25)
    q3 = clustered_data['Anomaly_Score'].quantile(0.75)
    iqr = q3 - q1
    threshold = q3 + 1.5 * iqr
    
    anomalies = clustered_data[clustered_data['Anomaly_Score'] > threshold]
    print(f"\n检测到{len(anomalies)}个异常VAV单元:")
    
    # 可视化异常单元
    plt.figure(figsize=(12,6))
    sns.scatterplot(data=clustered_data, x='AirFlow_mean', y='Temp_deviation',
                    hue='Cluster', size='Anomaly_Score', palette='viridis')
    plt.scatter(anomalies['AirFlow_mean'], anomalies['Temp_deviation'], 
                marker='X', s=200, color='red', label='Anomalies')
    plt.title('Anomaly Detection in Clusters')
    plt.legend()
    plt.show()
    
    return anomalies

# 执行异常检测
anomalous_units = identify_anomalies(clustered_data)
```

```python

# =============================================
# 第四部分：结果整合与输出
# =============================================

# 1. 保存预测模型
import joblib
joblib.dump(predictive_models, '../../outputs/HVAC_NewData_Analysis_Results/4.hvac_predictive_models.pkl')

# 2. 保存聚类结果
clustered_data.to_csv('../../outputs/HVAC_NewData_Analysis_Results/4.vav_cluster_analysis.csv')

# 3. 生成维修建议报告
anomalous_units.sort_values('Anomaly_Score', ascending=False).to_csv('../../outputs/HVAC_NewData_Analysis_Results/4.anomaly_units_priority.csv')

print("\n分析完成！关键结果已保存：")
print("- hvac_predictive_models.pkl (预测模型)")
print("- vav_cluster_analysis.csv (聚类分析)")
print("- anomaly_units_priority.csv (维修优先级)")
```
